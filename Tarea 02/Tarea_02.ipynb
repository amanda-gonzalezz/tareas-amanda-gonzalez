```python
# Problema 1: Clasificador de temperatura

# Recorre los números desde -30 hasta 50
for numero in range(-30, 51):
    if numero < 0:
        print(numero, "Muy frío")
    elif numero < 25:
        print(numero, "Templado")
    else:
        print(numero, "Caluroso")
```

```python
# Problema 2: Números especiales del robot

# Recorre los números del 1 al 100
for numero in range(1, 101):
# Comprobar si el número es divisible por 10
    if numero % 10 == 0:
        print(numero, "Bip Bip")
# Comprobar si el número es divisible por 5
    elif numero % 5 == 0:
        print(numero, "Bip")
```

```python
#Problema 3: Detector de puntaje gamer

# Pedir dos puntajes al usuario
puntaje1 = int(input("Ingresa primer puntaje: "))
puntaje2 = int(input("Ingresa segundo puntaje: "))

# Suma los dos puntajes
suma = puntaje1 + puntaje2

# Muestra el nivel según el resultado de la suma
if suma < 50:
    print("Resultado: Nivel principiante")
elif suma < 80:
    print("Resultado: Nivel intermedio")
else:
    print("Resultado: Nivel pro")
```

```python
# Problema 4: Calculador de perfil musical

# Definir la edad y los gustos de la persona
edad = 20
le_gusta_musica = True
le_gusta_bailar = False

# Evaluar el perfil según la edad y los gustos
if edad >= 18 and le_gusta_musica and le_gusta_bailar:
    print("Eres una estrella de la fiesta")
elif le_gusta_musica and not le_gusta_bailar:
    print("Te gusta la música, pero no tanto bailar")
else:
    print("Tienes tu propio estilo")
```
